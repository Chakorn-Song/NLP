```python
import pandas as pd
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
# import dataset
sms_test = pd.read_csv("SMS_test.csv", encoding='latin1')
sms_train = pd.read_csv("SMS_train.csv", encoding='latin1')
print(sms_test.head())
print(sms_train.head())
sms_train.columns

# rename columns
sms_test = sms_test.rename(columns={
    'S. No.': 'id',
    'Message_body': 'message',
    'Label':'label'
})
sms_train = sms_train.rename(columns={
    'S. No.': 'id',
    'Message_body': 'message',
    'Label':'label'
})
```

```python
tokenizer = Tokenizer(num_words = 100, oov_token = "<OOV>")
tokenizer.fit_on_texts(sms_train['Message_body']) 
word_index = tokenizer.word_index
print(word_index)

sequences = tokenizer.texts_to_sequences(sms_train['Message_body'])

padded = pad_sequences(sequences)
print(padded)
```
